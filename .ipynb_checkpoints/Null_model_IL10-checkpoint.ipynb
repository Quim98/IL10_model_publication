We generate a linear model using the same initial conditions as our IL-2 signaling model (RA_0, RB_0, RG_0, STAT5_0 and IL2_0) as predictors, trained in the same data as our IL-2 signaling model and tested in the cytokine dictionary data

```python
from sklearn.linear_model import LinearRegression
from scipy.optimize import curve_fit
from src.simulate_func import set_simulation,simulate,func_time,hill_func
import pandas as pd
import numpy as np
import seaborn as sns
from scipy.stats import spearmanr,pearsonr
import matplotlib.pyplot as plt
#from matplotlib import use
#use('Agg')
import warnings
warnings.filterwarnings('ignore')
plt.rc('xtick', labelsize=14)
plt.rc('ytick', labelsize=14)
```

```python
# Import datasets
df_IC_data = pd.read_csv('data/expression/whole_dataset.csv')
df_linear_models = pd.read_csv('data/lin_models/linear_models.csv')
df_sim_data = pd.read_csv('data/signaling/IL2_data.csv')
df_bind = pd.read_csv('data/binding/IL2_data_param.csv')
df_EC50_AMP = pd.read_csv('data/signaling/IL2_data_hill.csv')

cell_list_train = ["T8","T8.naive","Tregs","NK","T4_IL2RAlow","ACH-000061"]
cell_list_test = ["T8.M", "T8.naive", "T4.M_IL2RAlow", "T4.naive_IL2RAlow", "mTregs", "nTregs","ACH-000061"]

# Divide data into train and tests datasets as done in the IL-2 signaling model
df_sim_data_train = df_sim_data.loc[df_sim_data["Cell_type"].isin(cell_list_train)]
df_sim_data_train = df_sim_data_train.reset_index()[df_sim_data_train.columns]
df_EC50_AMP_train = df_EC50_AMP.loc[df_EC50_AMP["Cell_type"].isin(cell_list_train)]
df_EC50_AMP_train = df_EC50_AMP_train.reset_index()[df_EC50_AMP_train.columns]
df_sim_data_test = df_sim_data.loc[(df_sim_data["Cell_type"].isin(cell_list_test) & (df_sim_data["Variant"]=="WT"))]
df_sim_data_test = df_sim_data_test.reset_index()[df_sim_data_test.columns]
df_EC50_AMP_test = df_EC50_AMP.loc[(df_EC50_AMP["Cell_type"].isin(cell_list_test) & (df_EC50_AMP["Variant"]=="WT"))]
df_EC50_AMP_test = df_EC50_AMP_test.reset_index()[df_EC50_AMP_test.columns]
```

```python
# Save all data of models in a dataframe
df_models = pd.DataFrame(columns=["Model","r2_train","EC50e_train","AMPe_train","r2_test","EC50e_test","AMPe_test","SpearmanR","PearsonR"])
```

# Linear model only with receptor,STAT5 expression and IL2 concentration on multiple variants

```python
# As with IL-2 signaling model, get the number of receptors and STAT proteins per cell type (training data)
RA0_train = []
RB0_train = []
RG0_train = []
STAT50_train = []
for index in df_sim_data_train.index:
    df_cell = df_IC_data.loc[df_IC_data["Cell"]==df_sim_data_train.loc[[index],"Cell_type"].values[0]]
    if np.isnan(df_cell.loc[df_cell["Gene"]=="IL2RA_HUMAN"]["Log10 Prot. count"].values[0]):
        RA0_train.append((df_linear_models.loc[df_linear_models["Gene"]=="IL2RA_HUMAN"]["Intercept"].values[0] + df_cell.loc[df_cell["Gene"]=="IL2RA_HUMAN"]["Log2 TPM"].values[0] * df_linear_models.loc[df_linear_models["Gene"]=="IL2RA_HUMAN"]["Slope"].values[0]))
    else:
        RA0_train.append(df_cell.loc[df_cell["Gene"]=="IL2RA_HUMAN"]["Log10 Prot. count"].values[0])
    if np.isnan(df_cell.loc[df_cell["Gene"]=="IL2RB_HUMAN"]["Log10 Prot. count"].values[0]):
        RB0_train.append((df_linear_models.loc[df_linear_models["Gene"]=="IL2RB_HUMAN"]["Intercept"].values[0] + df_cell.loc[df_cell["Gene"]=="IL2RB_HUMAN"]["Log2 TPM"].values[0] * df_linear_models.loc[df_linear_models["Gene"]=="IL2RB_HUMAN"]["Slope"].values[0]))
    else:
        RB0_train.append(df_cell.loc[df_cell["Gene"]=="IL2RB_HUMAN"]["Log10 Prot. count"].values[0])
    if np.isnan(df_cell.loc[df_cell["Gene"]=="IL2RG_HUMAN"]["Log10 Prot. count"].values[0]):
        RG0_train.append((df_linear_models.loc[df_linear_models["Gene"]=="IL2RG_HUMAN"]["Intercept"].values[0] + df_cell.loc[df_cell["Gene"]=="IL2RG_HUMAN"]["Log2 TPM"].values[0] * df_linear_models.loc[df_linear_models["Gene"]=="IL2RG_HUMAN"]["Slope"].values[0]))
    else:
        RG0_train.append(df_cell.loc[df_cell["Gene"]=="IL2RG_HUMAN"]["Log10 Prot. count"].values[0])
    if np.isnan(df_cell.loc[df_cell["Gene"]=="STA5A_HUMAN"]["Log10 Prot. count"].values[0] + df_cell.loc[df_cell["Gene"]=="STA5B_HUMAN"]["Log10 Prot. count"].values[0]):
        STAT50_train.append(np.log10(10**(df_linear_models.loc[df_linear_models["Gene"]=="STA5A_HUMAN"]["Intercept"].values[0] + df_cell.loc[df_cell["Gene"]=="STA5A_HUMAN"]["Log2 TPM"].values[0] * df_linear_models.loc[df_linear_models["Gene"]=="STA5A_HUMAN"]["Slope"].values[0]) + 10**(df_linear_models.loc[df_linear_models["Gene"]=="STA5B_HUMAN"]["Intercept"].values[0] + df_cell.loc[df_cell["Gene"]=="STA5B_HUMAN"]["Log2 TPM"].values[0] * df_linear_models.loc[df_linear_models["Gene"]=="STA5B_HUMAN"]["Slope"].values[0])))
    else:
        STAT50_train.append(np.log10(10**(df_cell.loc[df_cell["Gene"]=="STA5A_HUMAN"]["Log10 Prot. count"].values[0]) + 10**(df_cell.loc[df_cell["Gene"]=="STA5B_HUMAN"]["Log10 Prot. count"].values[0])))
# Organize data to train linear model
df_sim_train = pd.DataFrame()
df_sim_train["RA0"] = RA0_train
df_sim_train["RB0"] = RB0_train
df_sim_train["RG0"] = RG0_train
df_sim_train["STAT50"] = STAT50_train
df_sim_train["IL0"] = np.log10(df_sim_data_train["IL"].values)
```

```python
# Train linear model and get r2 in training data
lin_model_full = LinearRegression().fit(df_sim_train.values, df_sim_data_train["pSTAT"].values)
print("r2 in training data: "+str(round(lin_model_full.score(df_sim_train.values, df_sim_data_train["pSTAT"].values),2)))
r2_train = lin_model_full.score(df_sim_train.values, df_sim_data_train["pSTAT"].values)
```

```python
# Get error in EC50 and amplitude in training data
num_sim = 20
df_sim_train_full = set_simulation(df_bind, df_sim_data_train, df_IC_data, df_linear_models, num_sim)
df_sim_train_full = df_sim_train_full[df_sim_train.columns]
df_sim_train_full["IL0"] = np.log10(df_sim_train_full["IL0"])
for plot_num in df_sim_data_train["Plot"].drop_duplicates():
    for variant in df_sim_data_train.loc[df_sim_data_train["Plot"]==plot_num,"Variant"].drop_duplicates():
        pSTAT5 = lin_model_full.predict(df_sim_train_full.loc[df_sim_data_train.loc[(df_sim_data_train["Plot"]==plot_num) & (df_sim_data_train["Variant"]==variant)].index].values)
        IL2 = 10**df_sim_train_full.loc[df_sim_data_train.loc[(df_sim_data_train["Plot"]==plot_num) & (df_sim_data_train["Variant"]==variant)].index,"IL0"].values
        fit, cov = curve_fit(hill_func, IL2, pSTAT5, bounds = ([min(pSTAT5),min(IL2)/10], [max(pSTAT5)+10, max(IL2)*10]))
        df_EC50_AMP_train.loc[(df_EC50_AMP_train["Plot"]==plot_num)&(df_EC50_AMP_train["Variant"]==variant),"EC50_m"] = np.log10(fit[1])
        df_EC50_AMP_train.loc[(df_EC50_AMP_train["Plot"]==plot_num)&(df_EC50_AMP_train["Variant"]==variant),"Amp_m"] = fit[0]
# Save errors
eEC50_train = (df_EC50_AMP_train["EC50_t"]-df_EC50_AMP_train["EC50_m"]).abs().mean()
eAMP_train = (df_EC50_AMP_train["Amp_t"]-df_EC50_AMP_train["Amp_m"]).abs().mean()
print("MAE_EC50: "+str(round(eEC50_train,2)))
print("MAE_Amplitude: "+str(round(eAMP_train,2)))
```

```python
# As with IL-2 signaling model, get the number of receptors and STAT proteins per cell type (test data)
RA0_test = []
RB0_test = []
RG0_test = []
STAT50_test = []
for index in df_sim_data_test.index:
    df_cell = df_IC_data.loc[df_IC_data["Cell"]==df_sim_data_test.loc[[index],"Cell_type"].values[0]]
    if np.isnan(df_cell.loc[df_cell["Gene"]=="IL2RA_HUMAN"]["Log10 Prot. count"].values[0]):
        RA0_test.append((df_linear_models.loc[df_linear_models["Gene"]=="IL2RA_HUMAN"]["Intercept"].values[0] + df_cell.loc[df_cell["Gene"]=="IL2RA_HUMAN"]["Log2 TPM"].values[0] * df_linear_models.loc[df_linear_models["Gene"]=="IL2RA_HUMAN"]["Slope"].values[0]))
    else:
        RA0_test.append(df_cell.loc[df_cell["Gene"]=="IL2RA_HUMAN"]["Log10 Prot. count"].values[0])
    if np.isnan(df_cell.loc[df_cell["Gene"]=="IL2RB_HUMAN"]["Log10 Prot. count"].values[0]):
        RB0_test.append((df_linear_models.loc[df_linear_models["Gene"]=="IL2RB_HUMAN"]["Intercept"].values[0] + df_cell.loc[df_cell["Gene"]=="IL2RB_HUMAN"]["Log2 TPM"].values[0] * df_linear_models.loc[df_linear_models["Gene"]=="IL2RB_HUMAN"]["Slope"].values[0]))
    else:
        RB0_test.append(df_cell.loc[df_cell["Gene"]=="IL2RB_HUMAN"]["Log10 Prot. count"].values[0])
    if np.isnan(df_cell.loc[df_cell["Gene"]=="IL2RG_HUMAN"]["Log10 Prot. count"].values[0]):
        RG0_test.append((df_linear_models.loc[df_linear_models["Gene"]=="IL2RG_HUMAN"]["Intercept"].values[0] + df_cell.loc[df_cell["Gene"]=="IL2RG_HUMAN"]["Log2 TPM"].values[0] * df_linear_models.loc[df_linear_models["Gene"]=="IL2RG_HUMAN"]["Slope"].values[0]))
    else:
        RG0_test.append(df_cell.loc[df_cell["Gene"]=="IL2RG_HUMAN"]["Log10 Prot. count"].values[0])
    if np.isnan(df_cell.loc[df_cell["Gene"]=="STA5A_HUMAN"]["Log10 Prot. count"].values[0] + df_cell.loc[df_cell["Gene"]=="STA5B_HUMAN"]["Log10 Prot. count"].values[0]):
        STAT50_test.append(np.log10((df_linear_models.loc[df_linear_models["Gene"]=="STA5A_HUMAN"]["Intercept"].values[0] + df_cell.loc[df_cell["Gene"]=="STA5A_HUMAN"]["Log2 TPM"].values[0] * df_linear_models.loc[df_linear_models["Gene"]=="STA5A_HUMAN"]["Slope"].values[0]) + 10**(df_linear_models.loc[df_linear_models["Gene"]=="STA5B_HUMAN"]["Intercept"].values[0] + df_cell.loc[df_cell["Gene"]=="STA5B_HUMAN"]["Log2 TPM"].values[0] * df_linear_models.loc[df_linear_models["Gene"]=="STA5B_HUMAN"]["Slope"].values[0])))
    else:
        STAT50_test.append(np.log10((df_cell.loc[df_cell["Gene"]=="STA5A_HUMAN"]["Log10 Prot. count"].values[0]) + 10**(df_cell.loc[df_cell["Gene"]=="STA5B_HUMAN"]["Log10 Prot. count"].values[0])))

# Organize data to test linear model
df_sim_test = pd.DataFrame()
df_sim_test["RA0"] = RA0_test
df_sim_test["RB0"] = RB0_test
df_sim_test["RG0"] = RG0_test
df_sim_test["STAT50"] = STAT50_test
df_sim_test["IL0"] = np.log10(df_sim_data_test["IL"].values) 
```

```python
print("r2 in test data: "+str(round(lin_model_full.score(df_sim_test.values, df_sim_data_test["pSTAT"].values),2)))
r2_test = lin_model_full.score(df_sim_test.values, df_sim_data_test["pSTAT"].values)
```

```python
# Get error in EC50 and amplitude in test data
num_sim = 20
df_sim_test_full = set_simulation(df_bind, df_sim_data_test, df_IC_data, df_linear_models, num_sim)
df_sim_test_full = df_sim_test_full[df_sim_test.columns]
df_sim_test_full["IL0"] = np.log10(df_sim_test_full["IL0"])
for plot_num in df_sim_data_test["Plot"].drop_duplicates():
    for variant in df_sim_data_test.loc[df_sim_data_test["Plot"]==plot_num,"Variant"].drop_duplicates():
        pSTAT5 = lin_model_full.predict(df_sim_test_full.loc[df_sim_data_test.loc[(df_sim_data_test["Plot"]==plot_num) & (df_sim_data_test["Variant"]==variant)].index].values)
        IL2 = 10**df_sim_test_full.loc[df_sim_data_test.loc[(df_sim_data_test["Plot"]==plot_num) & (df_sim_data_test["Variant"]==variant)].index,"IL0"].values
        fit, cov = curve_fit(hill_func, IL2, pSTAT5, bounds = ([min(pSTAT5),min(IL2)/10], [max(pSTAT5)+10, max(IL2)*10]))
        df_EC50_AMP_test.loc[(df_EC50_AMP_test["Plot"]==plot_num)&(df_EC50_AMP_test["Variant"]==variant),"EC50_m"] = np.log10(fit[1])
        df_EC50_AMP_test.loc[(df_EC50_AMP_test["Plot"]==plot_num)&(df_EC50_AMP_test["Variant"]==variant),"Amp_m"] = fit[0]
# Save errors
eEC50_test = (df_EC50_AMP_test["EC50_t"]-df_EC50_AMP_test["EC50_m"]).abs().mean()
eAMP_test = (df_EC50_AMP_test["Amp_t"]-df_EC50_AMP_test["Amp_m"]).abs().mean()
print("MAE_EC50: "+str(round(eEC50_test,2)))
print("MAE_Amplitude: "+str(round(eAMP_test,2)))
```

```python
# Get simulations of IL-2 signaling model to extract RA0, RB0, RG0, STAT50, IL0
df_sim_cytk_dict = pd.read_csv("results/immune_dict/simulations_cytk_dict.csv")
df_cytk_dict = pd.read_excel("data/signaling/41586_2023_6816_MOESM13_ESM.xlsx", sheet_name='Fig1d')
df_cytk_dict = df_cytk_dict.loc[df_cytk_dict["Cytokine"]=="IL-2"][["Celltype", "Magnitude", "Magnitude_norm"]]
df_cytk_dict = df_cytk_dict.loc[df_cytk_dict["Celltype"] != "Langerhans"]
df_cytk_dict = df_cytk_dict.loc[df_cytk_dict["Celltype"] != "Mast cell"]
df_cytk_dict["Celltype"] = ["B_cell", "T_cell_CD4", "T_cell_CD8", "T_cell_gd", "Treg", "NK_cell", "ILC", "pDC", "cDC1", "cDC2", "MigDC", "eTAC", "Macrophage", "Monocyte", "Neutrophil"]

# Organize data to for linear model
cells = df_sim_cytk_dict["Cell_type"]
df_sim_cytk_dict = df_sim_cytk_dict[['RA0', 'RB0', 'RG0', 'STAT50', 'IL0']]
IL0_cytk_dict = np.log10(df_sim_cytk_dict["IL0"].values)
df_sim_cytk_dict["RA0"] = (df_sim_cytk_dict["RA0"].values-np.mean(RA0_train))/np.std(RA0_train)
df_sim_cytk_dict["RB0"] = (df_sim_cytk_dict["RB0"].values-np.mean(RB0_train))/np.std(RB0_train)
df_sim_cytk_dict["RG0"] = (df_sim_cytk_dict["RG0"].values-np.mean(RG0_train))/np.std(RG0_train)
df_sim_cytk_dict["STAT50"] = (df_sim_cytk_dict["STAT50"].values-np.mean(STAT50_train))/np.std(STAT50_train)
df_sim_cytk_dict["IL0"] = (IL0_cytk_dict-np.mean(IL0_cytk_dict))/np.std(IL0_cytk_dict)

# Get linear model predictions and compare them with the experimental data
df_sim_cytk_dict["Result"] = lin_model_full.predict(df_sim_cytk_dict.values)
df_sim_cytk_dict["Cell_type"] = cells
df_sim_cytk_dict = df_sim_cytk_dict[["Cell_type","RA0","RB0","RG0","STAT50","Result"]]
df_sim_cytk_dict.columns = ["Celltype","RA0","RB0","RG0","STAT50","Result"]
df_sim_cytk_dict["Celltype"] = [cell.split("__")[0] for cell in df_sim_cytk_dict["Celltype"]]
df = pd.merge(df_cytk_dict, df_sim_cytk_dict, on='Celltype')
df.loc[df["Magnitude_norm"] == '≥100',"Magnitude_norm"] = str(100)
df["Magnitude_norm"] = df["Magnitude_norm"].astype("double")
df["Magnitude"] = df["Magnitude"].astype("double")
df["Result_log10"] = df["Result"]/df["Result"].max()*100
```

```python
print("Spearman's correlation coefficient: "+str(round(spearmanr(df["Result"], df["Magnitude_norm"])[0],2)))
rs = spearmanr(df["Result"], df["Magnitude_norm"])[0]
print("Pearson's correlation coefficient: "+str(round(pearsonr(df["Result"], df["Magnitude_norm"])[0],2)))
rp = pearsonr(df["Result"], df["Magnitude_norm"])[0]
```
